```python
import sqlite3

# Conecta a um banco de dados SQLite em memória (ou um arquivo, por exemplo, 'meu_banco.db')
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Comando SQL
sql_command = """
create table clientes(
    id INT,
    nome VARCHAR(50),
    cidade VARCHAR(50),
    idade INT
);
"""

# Executa o comando SQL
try:
    cursor.execute(sql_command)
    print("Tabela 'clientes' criada com sucesso.")
except sqlite3.Error as e:
    print(f"Erro ao criar a tabela: {e}")
finally:
    # Garante que a conexão seja fechada, mesmo se houver um erro
    conn.close()
```

```python
import sqlite3

# Conecta a um novo banco de dados SQLite em memória
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Recria a tabela 'clientes' (necessário pois o banco anterior foi fechado e destruído)
sql_create_table = """
create table clientes(
    id INT,
    nome VARCHAR(50),
    cidade VARCHAR(50),
    idade INT
);
"""
try:
    cursor.execute(sql_create_table)
    print("Tabela 'clientes' recriada para consulta.")
except sqlite3.Error as e:
    print(f"Erro ao recriar a tabela para consulta: {e}")

# Re-insere os dados (necessário pois o banco anterior foi fechado e destruído)
sql_insert_data = """
INSERT INTO clientes VALUES
(1, 'Ana', 'São Paulo', 25),
(2, 'Bruno', 'Rio de Janeiro', 30),
(3, 'Carla', 'São Paulo', 22),
(4, 'Diego', 'Belo Horizonte', 28),
(5, 'Eva', 'São Paulo', 35);
"""
try:
    cursor.execute(sql_insert_data)
    conn.commit()
    print("Dados inseridos para consulta.")
except sqlite3.Error as e:
    print(f"Erro ao inserir dados para consulta: {e}")

# Comando SQL de seleção
sql_select_data = """
SELECT * FROM clientes;

"""

# Executa o comando SQL de seleção
try:
    cursor.execute(sql_select_data)
    results = cursor.fetchall()

    # Imprime os resultados
    print("Dados na tabela 'clientes':")
    for row in results:
        print(row)
except sqlite3.Error as e:
    print(f"Erro ao selecionar dados: {e}")
finally:
    # Garante que a conexão seja fechada
    conn.close()
```

```python
import sqlite3

# Conecta a um novo banco de dados SQLite em memória
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Recria a tabela 'clientes' (necessário pois o banco anterior foi fechado e destruído)
sql_create_table = """
create table clientes(
    id INT,
    nome VARCHAR(50),
    cidade VARCHAR(50),
    idade INT
);
"""
try:
    cursor.execute(sql_create_table)
    print("Tabela 'clientes' recriada para agrupamento.")
except sqlite3.Error as e:
    print(f"Erro ao recriar a tabela para agrupamento: {e}")

# Re-insere os dados (necessário pois o banco anterior foi fechado e destruído)
sql_insert_data = """
INSERT INTO clientes VALUES
(1, 'Ana', 'São Paulo', 25),
(2, 'Bruno', 'Rio de Janeiro', 30),
(3, 'Carla', 'São Paulo', 22),
(4, 'Diego', 'Belo Horizonte', 28),
(5, 'Eva', 'São Paulo', 35);
"""
try:
    cursor.execute(sql_insert_data)
    conn.commit()
    print("Dados inseridos para agrupamento.")
except sqlite3.Error as e:
    print(f"Erro ao inserir dados para agrupamento: {e}")

# Comando SQL de seleção com agrupamento
sql_group_by_data = """
SELECT cidade, COUNT(*) AS total_clientes
FROM clientes
GROUP BY cidade;
"""

# Executa o comando SQL de seleção
try:
    cursor.execute(sql_group_by_data)
    results = cursor.fetchall()

    # Imprime os resultados
    print("Total de clientes por cidade:")
    for row in results:
        print(row)
except sqlite3.Error as e:
    print(f"Erro ao selecionar dados agrupados: {e}")
finally:
    # Garante que a conexão seja fechada
    conn.close()
```
